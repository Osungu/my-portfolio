# Functions

In Python, a function is a named block of reusable code that performs a specific task. It allows you to organize your code into logical units, making it easier to read, understand, and maintain. Functions help promote code reusability and modularity by encapsulating a set of instructions that can be called multiple times from different parts of a program.

1. Function Definition:

Functions are defined using the def keyword followed by the function name and parentheses.
The function name should follow Python naming conventions (lowercase letters and underscores for multi-word names).
Parentheses can optionally contain parameters (inputs) that the function expects.

2. Function Call:

To execute a function, you call it by using its name followed by parentheses.
If the function has parameters, you provide the corresponding arguments (values) within the parentheses.

3. Code Block:

A function consists of a code block that contains statements to be executed when the function is called.
The code block is indented under the function definition.
The indentation (typically four spaces) is crucial in Python to define the scope of the function.

4. Return Statement:

Functions can optionally return a value using the return statement.
The return statement specifies the value to be returned and ends the function's execution.
If no return statement is provided, the function returns None by default.

5. Function Parameters and Arguments:

Parameters are placeholders for values that a function expects to receive.
When calling a function, you provide arguments (actual values) that match the function's parameters.
Parameters can have default values, making them optional when calling the function.

Syntax:

def function_name(parameter1, parameter2, ...):
    # Function body (code block)
    # Statements to be executed when the function is called
    # ...
    [return value]

#### Creating functions

Creating a simple function

```python
def greet():
    print("Hello world!")

greet()  # Function call
```

Function with default parameters

```python
def greet(name="Friend"):
    print("Hello, " + name + "!")

greet()  # Output: Hello, Friend!
greet("Alice")  # Output: Hello, Alice!
```

#### Using the *args

    *args allows a function to accept any number of positional arguments.
    

```python
# if you pass more than one argument for a function that expects one argument, leads to an error


def perform_operations(args):
    print(args)

perform_operations(1,2,4)
```

```python
# to avoid getting an error, you use *arg to indicate that you will be passing more than one argument

def perform_operations(*args):
    print(args)

perform_operations(1,2,4)
```

```python
def calculate_sum(*args):
    total = 0
    for num in args:
        total += num
    return total

result = calculate_sum(1, 2, 3, 4, 5)
print("The sum is:", result)
```

#### Using the Keyword arguments

Keyword arguments in Python allow you to pass arguments to a function using the name of the parameter followed by the corresponding value. They provide flexibility and clarity when calling functions, as the arguments are explicitly identified by their parameter names. Here are detailed notes about keyword arguments in Python

Keyword arguments are passed using the key=value syntax when calling a function.

```python
def greet(name, message="Hello"):
    print(message + ", " + name + "!")
   
greet("victor")
greet(message="Bonjour", name="Jean")
```

 #### **kwargs

    **kwargs allows a function to accept any number of keyword arguments.

```python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(key + ": " + value)
        
print_details(name="Alice", age="25")
```

Combining *args and **kwargs

```python
import numpy as np

def perform_operations(*args, **kwargs):
    if 'sum' in kwargs.values():
        return np.sum(args)
    if 'multiply' in kwargs.values():
        return np.prod(args)

print(perform_operations(1, 2, 3, 4, operations='sum'))
print(perform_operations(1, 2, 3, 4, operations='multiply'))
```

#### The lambda functions


Lambda functions, also known as anonymous functions, are concise and inline functions in Python. They provide a way to create small, one-line functions without the need for a formal function definition. 

    Lambda functions are defined using the lambda keyword followed by a parameter list, a colon (:), and the expression to be evaluated.

```python
# lambda functions on a single digit

double = lambda x: x * 2
print(double(5))  # Output: 10
```

```python
#lambda functions on a double digit

numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
doubled
```
