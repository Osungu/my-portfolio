## Measures of Central Tendencies

```python
# creating Baseline and endline samples for beneficiaries and non-beneficiaries

Baseline= data[data["Program_level"]=="Baseline"]
Endline = data[data["Program_level"]=="Endline"]
BenefBaseline= Baseline[Baseline["BenefStatus"]=="Beneficiary"]
NonBenefBaseline= Baseline[Baseline["BenefStatus"]=="Non-Beneficiary"]
BenefEndline= Endline[Endline["BenefStatus"]=="Beneficiary"]
NonBenefEndline= Endline[Endline["BenefStatus"]=="Non-Beneficiary"]
```

```python
# Descriptive statistics for non-beneficiaries at baseline

NonBenefBaseline.describe()
```

```python
# Descriptive statistics for non-beneficiaries at endline

NonBenefEndline.describe()
```

##### Measures of central tendencies

```python
# Presenting the mean Age, mode Household size, Median Total Expenditure, Median Total income and Median debt, at endline

Endline.groupby("BenefStatus").agg({"AgeHH":"mean","HHsize":pd.Series.mode,"Total_Expenditure":"median","Total_Income":"median","Total_debt":"median"})
```

```python
# Presenting the mean Age, mode Household size, Median Total Expenditure, Median Total income and Median debt, at baseline

Baseline.groupby("BenefStatus").agg({"AgeHH":"mean","HHsize":pd.Series.mode,"Total_Expenditure":"median","Total_Income":"median","Total_debt":"median"})
```

## One Sample T-Test

```python
BenefEndline.describe()
```

```python
# Conducting the one sample T Test on Total expenditure
```

```python
import scipy.stats as stats
t_statistic, p_value = stats.ttest_1samp(a=BenefEndline['Total_Expenditure'], popmean=3000)
print(t_statistic , p_value)
```

```python
# Conducting the one sample T Test on Total Income
```

```python
t_statistic, p_value = stats.ttest_1samp(a=BenefEndline['Total_Income'], popmean=1600)
print(t_statistic , p_value)
```

```python
# perform one sample t-test on Total_debt
```

```python
t_statistic, p_value = stats.ttest_1samp(a=BenefEndline['Total_debt'], popmean=1500)
print(t_statistic , p_value)
```

# T-test with more than one sample 

### Repeated-measures t-test 

```python
# Extract the sample of beneficiaries only at Baseline
BenefBaseline= Baseline[Baseline["BenefStatus"]=="Beneficiary"]
```

```python
print(BenefBaseline["FCS"].mean(),BenefEndline["FCS"].mean())
```

```python
BenefBaseline["FCS"].mean()-BenefEndline["FCS"].mean()
```

```python
stats.ttest_rel(BenefBaseline["FCS"],BenefEndline["FCS"])
```

### The independent-groups t-test 

##### Getting the descriptive statistics of LCSI for beneficiaries and non beneficiaries at endline

```python
NonBenefEndline["LCSI"].describe()
```

```python
BenefEndline["LCSI"].describe()
```

##### Running the independent sample T-Test

```python
# running the independent sample T-Test
stats.ttest_ind(NonBenefEndline["LCSI"],BenefEndline["LCSI"], equal_var=True)
```
