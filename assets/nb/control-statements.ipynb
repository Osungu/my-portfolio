# Control Statements

Control statements, also known as control flow statements or flow control statements, are fundamental constructs in programming that allow you to control the flow of execution within a program. They determine the order in which statements are executed based on certain conditions or loops. Control statements enable you to make decisions, iterate over a set of values, and control the program's behavior.
There are three primary types of control statements:
   1. Conditional Statements such as if-else statements
   2. Loop Statements such as for and while
   3. Control Transfer Statements such as break and continue

### Conditional Statements

Conditional statements allow you to make decisions based on specified conditions. The most common conditional statement is the if-else statement, which executes a block of code if a certain condition is met, and a different block of code if the condition is not met.

#### The if - else statements

The if-else statement allows you to make decisions in your program based on specified conditions. It executes a block of code if the condition is True, and a different block of code if the condition is False. The else block is optional and can be omitted if there is no code to be executed when the condition is False.

    Syntax
        if condition:
            # code block if condition is True
        else:
            # code block if condition is False

```python
# Example 1

num = 9

if num % 2 == 0:
    print(f"{num} is an even number")
else:
    print(f"{num} is an odd number")
```

```python
# Example 2

age = 17

if age >= 18:
    print("You are eligible for enrolment into the program.")
else:
    print("You are not eligible for enrolment into the program.")
```

#### Elif Statements

elif statements (short for "else if") are used in conjunction with if statements to add additional conditional checks. They allow you to evaluate multiple conditions and execute specific code blocks based on the first condition that is true.

    Syntax
        if condition1:
            # code block executed if condition1 is true
        elif condition2:
            # code block executed if condition1 is false and condition2 is true
        elif condition3:
            # code block executed if both condition1 and condition2 are false and condition3 is true
        else:
            # code block executed if all conditions are false

```python
# Example 1

num = 10

if num > 0:
    print(f"{num} is a positive number")
elif num < 0:
    print(f"{num} is a negative number")
else:
    print(f"{num} is zero")
```

```python
# Example 2

score = 75

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
    

print("Your grade is", grade)
```

### Looping Statements

Loop Statements: Loop statements enable you to repeatedly execute a block of code until a specific condition is met. There are two main types of loop statements:

    For Loop: A for loop is used to iterate over a sequence (e.g., a list, tuple, string) or any iterable object. It executes a block of code for each item in the sequence.

    While Loop: A while loop repeats a block of code as long as a specified condition is True. It continually checks the condition before each iteration.

#### For Loops

The for loop in Python is used to iterate over a sequence (such as a list, tuple, string, or range) or any iterable object. It executes a block of code for each item in the sequence.

    Syntax
        for item in sequence:
            # code block executed for each item in the sequence

```python
# Example 1

fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

```python
# Example 2

numbers = [1, 2, 3, 4, 5]

sum = 0

for num in numbers:
    sum += num

print("Sum:", sum)
```

#### While loop

The while loop in Python is used to repeatedly execute a block of code as long as a specified condition is true. It is ideal when the number of iterations is unknown or based on a certain condition.

    Syntax:
        while condition:
            # code block executed as long as the condition is true

```python
# Example 1

count = 0

while count < 5:
    print(count)
    count += 1
```

```python
# Example 2

num = 10

while num > 0:
    print(num)
    num -= 2
```

### Control Transfer Statements

Control transfer statements allow you to alter the normal flow of a loop or program. break is used to exit the current loop, while continue is used to skip the current iteration and move to the next iteration of the loop.

#### Break Statements

The break statement is used to prematurely exit a loop when a certain condition is met. It is commonly used when you want to terminate a loop early based on a specific condition.

    Syntax:
        for item in sequence:
            if condition:
                break
    
    # code block executed before the loop is terminated

```python
# Example 1

for num in range(1, 6):
    if num == 4:
        break
    print(num)
```

break is a control transfer statement that, if the condition num == 4 is true, immediately exits the current loop. In this case, if num is equal to 4, the loop is terminated.

```python
# Example 2

fruits = ["apple", "banana", "cherry", "orange", "grape"]

for fruit in fruits:
    if fruit == "cherry":
        break
    print(fruit)
```

break is a control transfer statement that, if the condition fruit == "cherry" is true, immediately exits the current loop. In this case, if fruit is equal to cherry, the loop is terminated.

#### Continue statements

The continue statement is used to skip the current iteration of a loop and move to the next iteration. It is commonly used when you want to bypass a particular iteration based on a specific condition.

    Syntax:
         for item in sequence:
            if condition:
                continue
    
    # code block executed if the condition is false or after skipping the iteration       

```python
# Example 1

for num in range(1, 6):
    if num == 3:
        continue
    print(num)
```

```python
# Example 2

i = 0

while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
```
