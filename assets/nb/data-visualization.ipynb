##### 1. Pie Charts

###### Working Example for Pie Chart, Doughnut chart and  Grouped Bar Graph

```python
#importing the data

data = pd.read_csv("project satisfaction.csv")
```

```python
# exploring the first 5 rows
data.head()
```

```python
# Getting the frequency distributions of the satisfaction

pd.crosstab(index=data["Satisfaction"], columns="Frequency")
```

```python
# presenting a pie chart 

import matplotlib.pyplot as plt
labels = ["Very satisfied", "Satisfied", "Disatisfied", "Very dissatisfied"] #create a list of labels
sizes = [29, 90, 75, 29] # setting the pie sizes

# setting the chart area size as 5units on the horizontal and 10 units on the vertical axis
plt.figure(figsize=(5,10)) 

#plotting the pie chart
plt.pie(sizes,labels=labels,explode=[0.2,0,0,0],shadow=True,colors=['green','blue','yellow','red'],autopct="%2.1f%%") 

plt.title("Satisfaction with the project") 

plt.show()
```

##### 2. Doughnut chart

```python
# presenting the results in a Doughnut chart

import matplotlib.pyplot as plt
labels = ["Very satisfied", "Satisfied", "Disatisfied", "Very dissatisfied"] #setting labels
sizes = [29, 90, 75, 29] #setting the pie sizes

plt.figure(figsize=(5,10)) # setting the chart area size as 5units on the horizontal and 10 units on the vertical axis
plt.pie(sizes,labels=labels,explode=[0.1,0,0,0], colors=["red","blue","yellow","purple"],radius=1,autopct="%1.1f%%",pctdistance=0.85)

circle=plt.Circle((0,0),0.65,color="white") # the plt.Circle() function will draw a circle with center 0,0; radius = 0.65; with a color white

# plt.gcf stands for get current figure, plt.gcf() stands for get current axis.
# here we are getting the axis of the initial pie char, and adding the white circle inside it to create a doughnut chart
plt.gcf().gca().add_artist(circle)

plt.title("Satisfaction with the project")

plt.show()
```

##### 3. Grouped Bar Graph

```python
# get the frequency distribution of satisfaction by gender

pd.crosstab( data['Satisfaction'], data['Gender'], margins=True)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(layout="constrained")

male_bars = ax.bar([0,1,2,3], [22, 4, 58, 38], 0.25, label='Male')
ax.bar_label(male_bars, padding=3)

female_bars = ax.bar([0.25,1.25,2.25,3.25], [7, 86, 47, 37], 0.25, label='Female')
ax.bar_label(female_bars, padding=3)

total_bars = ax.bar([0.5,1.5,2.5,3.5], [29, 90, 105, 75], 0.25, label='Total')
ax.bar_label(total_bars, padding=3)

ax.set_xticks([0.25,1.25,2.25,3.25], ['Very satisfied','Satisfied','Disatisfied','Very disatisfied'])


ax.set_title('Project satisfaction by gender')
ax.set_ylabel('Number of beneficiaries')
ax.set_xlabel('Beneficiary satisfaction')

# setting the legend
ax.legend(loc='upper right',ncol=1, frameon=False, title='Gender')

plt.show()
```

##### 4. Stem plots

```python
import pandas as pd
import numpy as np

data= pd.read_csv("disease cases.csv")

import matplotlib.pyplot as plt

plt.figure(figsize=(13,5))
plt.stem(data["Date"], data["Diarrhea Cases"])
plt.xticks(rotation=45)  # Rotate x-axis labels for better visibility

plt.title("Diarrhea Cases in the Community Unit")
plt.ylabel("Diarrhea Cases")
plt.xlabel("Date")

plt.show()
```

### Line graph

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("disease cases.csv")

plt.figure(figsize=(13, 5))
plt.plot(data["Date"], data["Diarrhea Cases"], marker='o', linestyle='-', color='blue')

plt.title("Diarrhea Cases in the Community Unit")
plt.ylabel("Diarrhea Cases")
plt.xlabel("Date")

plt.xticks(rotation=45)  # Rotate x-axis labels for better visibility

plt.grid(True)  # Add gridlines

plt.show()
```

#### Epi-Curve

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(13,5))
plt.plot(data['Date'], data['Diarrhea Cases'], marker='o', linestyle='-',color='lightblue')
plt.fill_between(data['Date'], data['Diarrhea Cases'], color = 'lightblue')

plt.title('Diarrhea Cases in the Community Unit')
plt.ylabel('Diarrhea Cases')
plt.xlabel('Date')

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data=pd.read_csv("scatter.csv")
```

##### 5. Countplot

```python
import seaborn as sns
sns.countplot(x=data["gender"])
plt.title("Count of male and female respondents")
plt.show()
```

##### 6. Boxplot

```python
# we start by exploring the data
boxplot = data.boxplot(column='Age',by='gender')
```

##### 7. Histogram

```python
plt.hist(data[data['gender']=="female"]['Age'], bins = 5)
plt.title('Distribution of Age for female students')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.hist(data[data['gender']=="male"]['Age'], bins = 5)
plt.title('Distribution of Age for male students')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

##### 8. Distribution plots

```python
import seaborn as sns
sns.displot(data[data['gender']=="female"]['Age'], kind='kde')
plt.title("Distribution plot of gender")
plt.show()
```

```python
import seaborn as sns
sns.displot(data[data['gender']=="male"]['Age'], kind='kde')
plt.show()
```

##### 9. Pairplots

```python
import seaborn as sns
sns.pairplot(data=data, vars=['Age', 'math score', 'reading score', 'remedial lessons'], hue='gender')
```

##### 10. Correlation Matrix

```python
plt.figure(figsize=(8,8))

corr_matrix = data[['Age','math score','reading score', 'remedial lessons']].corr()

sns.heatmap(corr_matrix, annot = True)

# display the plot
plt.show()
```

##### 11. Scatter Plot

```python
# create a dummy variable representing color where "group A" is red, "group B" is black, "group C" is yellow, and "group D" is orange

data["colors"] = np.select([data["race/ethnicity"] == "group A",data["race/ethnicity"] =="group B",data["race/ethnicity"] == "group C",data["race/ethnicity"] == "group D"], ["red", "purple", "black", "green"])
```
